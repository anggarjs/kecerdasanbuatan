```python
#Written by Angga Rajasa
#NRP: 6022251114

import numpy as np
import matplotlib.pyplot as plt

# ── Fungsi segitiga
def triangle(x, a, b, c, d, e):
    if x < a:
        return d
    elif a <= x <= b:
        return (x - a) / (b - a)
    elif b < x <= c:
        return (c - x) / (c - b)
    elif x > c:
        return e
    else:
        return 0

#Input 1: Luasan area PV (0 sd 50 m2)
def fuzzifikasi_area(x):
    sedikit = triangle(x,  0,  0, 25, 1, 0)
    sedang  = triangle(x,  0, 25, 50, 0, 0)
    banyak  = triangle(x, 25, 50, 50, 0, 1)
    return np.array([sedikit, sedang, banyak])

# ── Input 2: Daya PV (0 sd 10 kWp)
def fuzzifikasi_daya(x):
    kecil    = triangle(x,  0,  0,  5, 1, 0)
    menengah = triangle(x,  0,  5, 10, 0, 0)
    besar    = triangle(x,  5, 10, 10, 0, 1)
    return np.array([kecil, menengah, besar])

#Rule table 3 × 3 (singleton dalam juta Rp)
#Kecil  Menengah  Besar
rule_FD = np.array([
    [5,   15,  15],   # Sedikit
    [5,   15,  30],   # Sedang
    [15,  30,  30],   # Banyak
])

#Input crisp
in_area = 10.0   # m²
in_daya = 7.0   # kWp

fuzzy_area = fuzzifikasi_area(in_area)
fuzzy_daya = fuzzifikasi_daya(in_daya)

#Inferensi + Defuzzifikasi
num   = 0
denum = 0
for i, val_area in enumerate(fuzzy_area):
    for j, val_daya in enumerate(fuzzy_daya):
        min_val = min(val_area, val_daya)   # AND = min()
        num     = num   + min_val * rule_FD[i][j]
        denum   = denum + min_val

crisp_output = num / denum

#Menentukan jenis paket terpilih
if crisp_output <= 10:
    paket = "Basic"
elif crisp_output <= 22.5:
    paket = "Premium"
else:
    paket = "Platinum"

#Output
print("Luasan area (m²)                     :", in_area)
print("Daya (kWp)                           :", in_daya)
print("fuzzy_area  (sedikit, sedang, banyak):", fuzzy_area)
print("fuzzy_daya  (kecil, menengah, besar) :", fuzzy_daya)
print("Harga paket (juta Rp)                :", crisp_output)
print("Paket terpilih                       :", paket)

```
